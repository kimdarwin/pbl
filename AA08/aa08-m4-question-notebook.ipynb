### **5. 질소 농도(ppm)(nutrient_n) 컬럼의 이상치를 IQR 기준으로 제거하고자 합니다. 다음 지침에 따라 처리하세요.**

<sub>학습 목표: 사분위 범위(IQR) 등의 기준으로 이상치를 찾아 처리할 수 있다 | 역량: DP1 이상치 처리 | 기술: pandas(quantile, reset_index, drop) | 난이도: 보통 | 예상 시간: 8분</sub>

- 앞 문항 결과 사용 : 2번 `df_raw`
- 대상 데이터프레임 : df_raw
- 사분위수(Q1, Q3)와 IQR 을 계산하세요
- lower_fence = Q1 - 3.0 x IQR, upper_fence = Q3 + 3.0 x IQR 로 정의하세요
- nutrient_n 이(가) 두 fence 밖에 있는 행을 삭제하세요
- 식별자 컬럼 serial_no 도 삭제하고 인덱스를 0부터 다시 매겨 df_clean 에 저장하세요
- 아래 코드 셀(cell)의 코드를 실행해 결과를 확인하세요 — 빈칸(____)은 한 곳입니다
- **[정답05]** 아래 코드 셀의 빈칸(____)에 들어갈 내용(함수·인자·모듈 이름)을 입력하세요

- 배점 : 6점

---

````python
# (코드 셀) 코드의 빈칸(____)을 채우고 실행하세요

q1 = df_raw['nutrient_n'].____(0.25)
q3 = df_raw['nutrient_n'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 3.0 * iqr
upper_fence = q3 + 3.0 * iqr

is_out = (df_raw['nutrient_n'] > upper_fence) | (df_raw['nutrient_n'] < lower_fence)
df_clean = df_raw[~is_out].drop(columns=['serial_no']).reset_index(drop=True)
````
