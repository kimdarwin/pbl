### **5. 누적 결제 횟수(PURCHASES) 컬럼의 이상치를 IQR 기준으로 제거하고자 합니다. 아래 안내대로 처리하십시오.**

<sub>학습 목표: 사분위 범위(IQR) 등의 기준으로 이상치를 찾아 처리할 수 있다 | 역량: DP1 이상치 처리 | 기술: pandas(quantile, reset_index, drop) | 난이도: 보통 | 예상 시간: 8분</sub>

- 앞 문항 결과 사용 : 2번 `data`

- 사분위수(Q1, Q3)와 IQR 을 산출하세요
- lower_fence = Q1 - 2.0 x IQR, upper_fence = Q3 + 2.0 x IQR 로 정의하세요
- PURCHASES 이(가) 두 fence 밖에 있는 행을 삭제하세요
- 식별자 컬럼 UID 도 삭제하고 인덱스를 0부터 다시 매겨 data_temp 에 보관하세요

````python
# (코드 셀) 코드의 빈칸(____)을 채우고 실행하세요

q1 = data['PURCHASES'].____(0.25)
q3 = data['PURCHASES'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 2.0 * iqr
upper_fence = q3 + 2.0 * iqr

is_out = (data['PURCHASES'] > upper_fence) | (data['PURCHASES'] < lower_fence)
data_temp = data[~is_out].drop(columns=['UID']).reset_index(drop=True)
````
