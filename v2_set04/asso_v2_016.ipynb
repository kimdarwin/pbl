### **5. 아래 코드는 상담 이관 횟수(Transfer_Cnt) 의 이상치(IQR 기준 K=2.0)와 식별자 컬럼을 제거하려는 코드인데 실행 중 예외가 발생합니다. 코드의 오류를 바로잡으세요.**

- (5-1) 코드의 오류를 정정하세요
    - 대상 데이터프레임 : df_raw
    - 결과 데이터프레임 : df_clean (이상치 행 삭제, 식별자 Record_Id 삭제, 인덱스 재설정)
- 배점 : 총 6점  [(5-1) 3점, (5-2) 3점]

---

````python
# (5-1) 여기에 코드의 오류를 정정하고 실행하세요

q1 = df_raw['Transfer_Cnt'].quantile(0.25)
q3 = df_raw['Transfer_Cnt'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 2.0 * iqr
upper_fence = q3 + 2.0 * iqr

df_clean = df_raw.drop(df_raw[(df_raw['Transfer_Cnt'] > upper_fence) | (df_raw['Transfer_Cnt'] < lower_fence)].index, axis=2)
df_clean = df_clean.drop(columns=['Record_Id'])
df_clean = df_clean.reset_index(drop=True)
````
