### **5. 그래프로 확인한 결과, 이전 편 지연(분)(delay_prev_min) 컬럼에서 이상치가 발견되었습니다. AI모델링할 때 부정적 영향이 없도록 제시된 가이드에 맞춰 이상치를 처리하세요.**

<sub>학습 목표: 사분위 범위(IQR) 등의 기준으로 이상치를 찾아 처리할 수 있다 | 역량: DP1 이상치 처리 | 기술: pandas(quantile, drop, reset_index) | 난이도: 쉬움 | 예상 시간: 6분</sub>

- 앞 문항 결과 사용 : 2번 `df`
- 대상 데이터프레임 : df
- 이전 편 지연(분)(delay_prev_min)이(가) upper fence 값보다 크거나 lower fence 값보다 작은 이상치 행(row)을 삭제하세요 (IQR 기준 K=2.0)
- AI모델링에 불필요한 표본 식별번호(sample_id) 컬럼을 삭제하세요
- 전처리를 적용한 데이터프레임은 df_temp 라는 변수명으로 담으세요
